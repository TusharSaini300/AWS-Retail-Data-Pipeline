# Gold Layer – Table Validation

This notebook checks the four Gold tables registered in the Glue Data Catalog.

**Database:** `retail_gold_db`

```sql
-- 1. List Gold tables
SHOW TABLES IN retail_gold_db;
```

```sql
-- 2. Compare Gold table row counts
SELECT 'dim_products' AS table_name, COUNT(*) AS total_rows
FROM retail_gold_db.dim_products
UNION ALL
SELECT 'dim_customers', COUNT(*)
FROM retail_gold_db.dim_customers
UNION ALL
SELECT 'dim_date', COUNT(*)
FROM retail_gold_db.dim_date
UNION ALL
SELECT 'fact_order_items', COUNT(*)
FROM retail_gold_db.fact_order_items;
```

```sql
-- 3. Find duplicate product IDs
SELECT product_id, COUNT(*) AS duplicate_count
FROM retail_gold_db.dim_products
GROUP BY product_id
HAVING COUNT(*) > 1
ORDER BY duplicate_count DESC;
```

```sql
-- 4. Find duplicate customer IDs
SELECT customer_id, COUNT(*) AS duplicate_count
FROM retail_gold_db.dim_customers
GROUP BY customer_id
HAVING COUNT(*) > 1;
```

```sql
-- 5. Find duplicate date IDs
SELECT date_id, COUNT(*) AS duplicate_count
FROM retail_gold_db.dim_date
GROUP BY date_id
HAVING COUNT(*) > 1;
```

```sql
-- 6. Check important fact-table nulls
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(transaction_id IS NULL) AS null_transaction_ids,
    COUNT_IF(customer_id IS NULL) AS null_customer_ids,
    COUNT_IF(product_id IS NULL) AS null_product_ids,
    COUNT_IF(date_id IS NULL) AS null_date_ids,
    COUNT_IF(net_amount IS NULL) AS null_net_amounts,
    COUNT_IF(net_amount_inr IS NULL) AS null_inr_amounts
FROM retail_gold_db.fact_order_items;
```

## How to interpret

- Duplicate checks should normally return zero rows for dimension keys.
- Null checks should be interpreted against the transformations and source data.
- A duplicate `product_id` is especially worth investigating because the Gold product-dimension joins can multiply rows if a brand maps to multiple category records.
