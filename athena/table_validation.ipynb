# Silver Layer – Table Validation

This notebook validates the Silver layer in the AWS Glue Data Catalog using Amazon Athena.

**Database:** `retail_silver_db`

Run the cells in order. The expected counts below are based on the project’s recorded Silver transformations; `order_items` count must be verified from the actual pipeline output.

```sql
# 1. List Silver tables
SHOW TABLES IN retail_silver_db;
```

```sql
-- 2. Validate row counts across all Silver tables
SELECT 'brands' AS table_name, COUNT(*) AS total_rows
FROM retail_silver_db.brands
UNION ALL
SELECT 'category', COUNT(*)
FROM retail_silver_db.category
UNION ALL
SELECT 'customers', COUNT(*)
FROM retail_silver_db.customers
UNION ALL
SELECT 'date', COUNT(*)
FROM retail_silver_db.date
UNION ALL
SELECT 'products', COUNT(*)
FROM retail_silver_db.products
UNION ALL
SELECT 'order_items', COUNT(*)
FROM retail_silver_db.order_items;
```

```sql
-- 3. Check customer nulls
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(customer_id IS NULL) AS null_customer_ids,
    COUNT_IF(phone IS NULL) AS null_phones
FROM retail_silver_db.customers;
```

```sql
-- 4. Check product nulls
SELECT
    COUNT(*) AS total_rows,
    COUNT_IF(product_id IS NULL) AS null_product_ids,
    COUNT_IF(brand_code IS NULL) AS null_brand_codes,
    COUNT_IF(category_code IS NULL) AS null_category_codes
FROM retail_silver_db.products;
```

```sql
-- 5. Inspect sample records
SELECT *
FROM retail_silver_db.products
LIMIT 20;
```

```sql
SELECT *
FROM retail_silver_db.order_items
LIMIT 20;
```
